```python
puzzle_input = """
..@@.@@@@.
@@@.@.@.@@
@@@@@.@.@@
@.@@@@..@.
@@.@@@@.@@
.@@@@@@@.@
.@.@.@.@@@
@.@@@.@@@@
.@@@@@@@@.
@.@.@@@.@.
"""
```

```python
split = puzzle_input.strip().split("\n")
split
```

```python
def is_roll(x, y):
    # Boundary checks first
    if x < 0 or y < 0:
        return False

    if x >= len(split[0]) or y >= len(split):
        return False
    try:
        return split[y][x] == "@"
    except IndexError:
        print(x, y)
```

```python
roll_counts = []

for y in range(len(split)):
    for x in range(len(split[0])):
        if is_roll(x, y):
            roll_counts.append([is_roll(x+i, y+j) for i in range(-1, 2) for j in range(-1, 2)])
```

```python
roll_counts
```

```python
neighbor_counts = [sum(counts) for counts in roll_counts]
```

```python
result = [x for x in neighbor_counts if x <= 4]
result
```
