```python
puzzle_input = """
aaa: you hhh
you: bbb ccc
bbb: ddd eee
ccc: ddd eee fff
ddd: ggg
eee: out
fff: out
ggg: out
hhh: ccc fff iii
iii: out
"""
```

```python
lines = puzzle_input.strip().split("\n")
lines
```

```python
cons = {}

for line in lines:
    [start, end] = line.split(": ")
    cons[start] = end.split(" ")


cons
```

```python
def bt(node, graph, cache, out):
    if node == out:
        return 1
    key = node
    if key in cache:
        return cache[key]
    total = 0
    for output in graph[node]:
        total += bt(output, graph, cache, out)
    cache[key] = total
    return total
```

```python
bt("you", cons, {}, "out")
```
