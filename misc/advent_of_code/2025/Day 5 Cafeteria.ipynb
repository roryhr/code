```python
puzzle_input = """
3-5
10-14
16-20
12-18

1
5
8
11
17
32
"""
```

```python
[first, second] = puzzle_input.strip().split("\n\n")
# first
```

```python
fresh_split = first.split("\n")
```

```python
# fresh = []

# for l in fresh_split:
#     [s, e] = l.split("-")
#     s, e = int(s), int(e)
#     fresh.append(list(range(s, e+1)))
```

```python
fresh = []

for l in fresh_split:
    [s, e] = l.split("-")
    s, e = int(s), int(e)
    fresh.append((s, e))
```

```python
fresh
```

```python
ingredients = list(map(int, second.split("\n")))
```

```python
ingredients
```

```python
def check_ingredient(i):
    for s, e in fresh:
        if s <= i <= e:
            return True
    return False
```

```python
len([x for x in ingredients if check_ingredient(x)])
```
